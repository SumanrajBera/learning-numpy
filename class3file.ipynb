## Numpy array operations

Lets first see how arrays behave like “vectors/matrices” when operations happen element-wise.

```python
import numpy as np

a = np.array([1,2,3,4])
b = np.array([21, 32, 40, 50])
print(a + b)
print(a * b)
print(a - b)
print(a / b) # Normal division
print(b // a) # Floor division
```

These things are already different from List and if we talk about one more special power it is **Broadcasting**

```python
arr = np.array([10, 20, 30, 40])
print(arr * 2)
print(arr + 2)
```

Even these things work with different shaped array the calculations will happen on every row. but dimensions must match

```python
arr2D = np.array([[10, 20, 30, 40], [50, 60, 70, 80]])
print(arr2D * np.array([5, 5, 5, 5])) # Gets multiplied row wise
print(arr2D * np.array([[5],[2]])) # Gets multiplied row and col wise
print(arr2D * 2) # This multiplies by a single value
```

# Aggregate Functions

These functions are used to calculate the statistics on arrays.

```python
stat1D = np.array([1, 5, 16, 2 , 20, 15])
print(np.sum(stat1D))
print(np.max(stat1D))
print(np.min(stat1D))
print(np.mean(stat1D))
print(np.std(stat1D))
print(np.var(stat1D))
```

You can also do the axis based operations row-Wise and Colum-wise

```python
stat2D = np.array([[10, 20, 30, 40], [50, 60, 70, 80]])
```

```python
print(np.sum(stat2D))
```

```python
# row and column wise using axis
print(np.sum(stat2D, axis=1)) # Along the x-axis
print(np.sum(stat2D, axis=0)) # Along the y-axis
```
