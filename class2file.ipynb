## 1. Attributes of an Array

Introduce attributes that describe the array.

Attributes: .shape, .ndim, .size, .dtype.

```python
import numpy as np
arr = np.random.randint(1,10, 5)
arr
```

```python
print(arr.shape)
print(arr.ndim)
print(arr.size)
print(arr.dtype)
```

```python
arr2D = np.random.randint(1,10, (3,3))
```

```python
arr2D
```

```python
print(arr2D.shape)
print(arr2D.ndim)
print(arr2D.size)
print(arr2D.dtype)
```

## Indexing and Slicing
Now we have Indexing and slicing aswell and for that we are going to see 1d and 2d indexing lets start with **1D indexing** and it is similar to list.

```python
arr[0]
```

Now lets see the 2D indexing aswell

```python
arr2D[1,2]
```

```python
arr2D[1]
```

Now we can also do the slicing lets see the **slicing in 1D**

```python
print(arr[:])
print(arr[0:3])
print(arr[0:5:2])
```

1D slicing was really easy now lets see the **slicing in 2D**

```python
print(arr2D[1:,1:])
print(arr2D[1:3,:])
```

There is one more thing known as **Boolean indexing**

```python
bool1D = np.array([1,4,2,9,3])
print(bool1D > 2)
print(bool1D[bool1D > 2])
```

```python
bool2D = np.random.randint(1,10, (4,4))
bool2D
```

```python
print(bool2D > 5)
print(bool2D[bool2D > 5])
```

## Reshaping the Array

you can also reshape the Array as you like but be accurate about the dimensions

```python
arr1 = np.arange(1,31).reshape(6,5)
print(arr1)
arr2 = np.array([[1,2,3],[4,5,6]])
print(arr2)
print(arr2.reshape(1,6))
print(arr2.reshape(3,2))
```

## Flattening of an array

you can convert the 2D array to 1D using the Flattening Function

```python
arr2.flatten()
```
