## Q3. Conditional selection

A company records employee performance scores.

```python
scores = np.array([45, 78, 92, 34, 67, 88, 51, 96, 72, 40])
```

Your tasks:

1. Retrieve all scores above 70.
2. Retrieve all scores below 50.
3. Find the indices of scores greater than or equal to 90.
4. Create a new array where scores below 50 become 0, while other scores remain unchanged.
5. Create a Boolean array identifying scores between 60 and 90, inclusive.


````python
scores = np.array([45, 78, 92, 34, 67, 88, 51, 96, 72, 40])

# Your solution
print(scores[scores > 70])
print(scores[scores < 50])
print(np.where(scores < 50, 0, scores))
print(np.where((scores >= 60) & (scores <= 90), True, False))
````

## Q4. Arithmetic and broadcasting

An e-commerce platform records product prices:

```python

    [100, 200, 300, 400],
    [150, 250, 350, 450],
    [200, 300, 400, 500]
])
```

Perform these operations:

1. Increase every product price by 10%.
2. Apply a flat discount of 20 to every product.
3. Apply a different discount to each product category: `[5, 10, 15, 20]`.
4. Apply different multipliers to each store: `[[1.1], [1.2], [1.3]]`.
5. Calculate the element-wise difference between the original and discounted prices.


````python
prices = np.array([
    [100, 200, 300, 400],
    [150, 250, 350, 450],
    [200, 300, 400, 500]
])

# Your solution
print(prices + (prices * 0.1))
print(prices - 20)
print(prices - (prices * np.array([0.05, 0.1, 0.15, 0.2])))
print(prices * np.array([[1.1], [1.2], [1.3]]))
discount = prices - (prices * np.array([0.05, 0.1, 0.15, 0.2]))
print(prices - discount)
````

## Q6. Fancy indexing

Consider this student marks dataset:

```python
marks = np.array([
    [78, 85, 90, 67],
    [88, 76, 92, 81],
    [65, 89, 73, 95],
    [91, 84, 79, 88]
])
```

Perform these operations:

1. Retrieve the marks at indices 0, 2 and 3 from the first row.
2. Retrieve these specific elements:
   - Row 0, Column 1
   - Row 2, Column 3
   - Row 3, Column 2
3. Retrieve the diagonal elements.
4. Retrieve the first and last rows, selecting only the second and fourth columns.


````python
marks = np.array([
    [78, 85, 90, 67],
    [88, 76, 92, 81],
    [65, 89, 73, 95],
    [91, 84, 79, 88]
])

# Your solution
print(marks[[0],[0,2,3]])
print(marks[0,1])
print(marks[2,3])
print(marks[3,2])
print(np.diag(marks))
print(marks[np.ix_([0,3],[1,3])])
````

## Q8. Concatenation and stacking

Two branches submit their monthly sales:

```python
branch_a = np.array([
    [100, 200, 300],
    [150, 250, 350]
])

branch_b = np.array([
    [400, 500, 600],
    [450, 550, 650]
])
```

Perform the following:

1. Combine both arrays into one 2D array by adding the rows of the second branch after the first.
2. Combine both arrays horizontally.
3. Combine both arrays vertically, with branch B appearing first.
4. Flatten the horizontally combined result.


````python
branch_a = np.array([
    [100, 200, 300],
    [150, 250, 350]
])

branch_b = np.array([
    [400, 500, 600],
    [450, 550, 650]
])

# Your solution
print(np.concatenate([branch_a, branch_b]))
print(np.hstack([branch_a, branch_b]))
print(np.concatenate([branch_b, branch_a]))
print(np.hstack([branch_a, branch_b]).flatten())
````
