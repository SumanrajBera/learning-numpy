## Fancy Indexing
you can Pick elements from specific positions using a list/array of indices.

```python
import numpy as np
arr = np.arange(1,20)
print(arr)
print(arr[[1,5, 8]]) # To get multiple index
```

You can do this fancy indexing in 2D Matrix aswell

```python
arr2D = np.arange(1,13).reshape(3, 4)

print(arr2D)
print(arr2D[1,2]) # To get a single element
print(arr2D[[0, 2, 2],[1,3, 2]]) # TO get multiple index in 2D matrix
```

## Conditional Selection
Find values based on conditions or replace them.

```python
arr = np.random.randint(1, 20, 5)

print(arr)
print(arr[arr > 5])
print(np.where(arr > 5)) # Based on condition only the index where its greater we get those index
print(np.where(arr > 5, arr, 0)) # (condition, when-true-become-arr-element, when-false-become-0) 
```

## Sorting & Ordering
np.sort() → returns sorted array

np.argsort() → returns indices of sorted order

```python
arr = np.random.randint(30, 60, 5)
print(arr)
```

```python
print(np.sort(arr))
print(np.argsort(arr)) # index of sorted order
```

## Concatenation & Stacking

```python
a = np.array([1,2,3,4])
b = np.array([5,6,7])
np.concat([a,b])
```

Vertical & Horizontal Stacking

```python
A = np.array([[1,2,3],[4,5,6]])
B = np.array([[7,8,9],[10,11,12]])
```

```python
np.vstack([B, A])
```

```python
np.hstack([A, B])
```
